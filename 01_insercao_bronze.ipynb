```sql
-- Camada Bronze: Inserção dos dados brutos da Olist
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
```

```sql
-- Validação de volumetria da Camada Bronze
SELECT 'orders' AS tabela, COUNT(*) AS total_linhas FROM bronze.bronze_orders
UNION ALL
SELECT 'order_items', COUNT(*) FROM bronze.bronze_order_items
UNION ALL
SELECT 'order_reviews', COUNT(*) FROM bronze.bronze_order_reviews
UNION ALL
SELECT 'products', COUNT(*) FROM bronze.bronze_products
UNION ALL
SELECT 'customers', COUNT(*) FROM bronze.bronze_customers;
```

```sql
ALTER TABLE bronze.bronze_customers RENAME TO bronze.customers;
ALTER TABLE bronze.bronze_order_items RENAME TO bronze.order_items;
ALTER TABLE bronze.bronze_orders RENAME TO bronze.orders;
ALTER TABLE bronze.bronze_order_reviews RENAME TO bronze.order_reviews;
ALTER TABLE bronze.bronze_products RENAME TO bronze.products;
```
